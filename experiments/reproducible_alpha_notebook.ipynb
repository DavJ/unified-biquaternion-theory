## Step 6: Prime Stability and n* = 137  [CONDITIONAL]

The effective potential $V_{\text{eff}}(n) = n^2 - B \ln n$ is minimised over prime integers.

**Prime stability theorem**: A winding vacuum at $n^*$ is topologically stable
if and only if $n^*$ is prime (no sub-harmonic modes exist for prime winding numbers).

**Non-circularity check**: We test multiple $N_{\text{eff}}$ values to verify that 137
is not always selected — only N_eff = 12 (Standard Model) gives n* = 137.

```python
# Helper functions: prime sieve and V_eff

def sieve_primes(limit: int) -> List[int]:
    """Sieve of Eratosthenes: return all primes up to limit."""
    if limit < 2:
        return []
    is_prime = bytearray([1]) * (limit + 1)
    is_prime[0] = is_prime[1] = 0
    for i in range(2, int(limit**0.5) + 1):
        if is_prime[i]:
            is_prime[i*i::i] = bytearray(len(is_prime[i*i::i]))
    return [i for i in range(2, limit + 1) if is_prime[i]]

PRIMES_600 = sieve_primes(600)

def V_eff(n: float, B: float) -> float:
    """Effective potential: V(n) = n² − B·n·ln(n)  [A=1 normalized]"""
    if n <= 0:
        return float('inf')
    return n**2 - B * n * math.log(n)

def prime_minimizer(B: float, primes: List[int] = PRIMES_600) -> Tuple[int, float]:
    """Return (p*, V(p*)) = argmin over primes of V_eff(p)."""
    best = min(primes, key=lambda p: V_eff(float(p), B))
    return best, V_eff(float(best), B)

def continuous_minimum(B: float) -> float:
    """Continuous minimum of V=n²−B·n·ln n: solve 2n = B(ln n + 1) numerically."""
    # Newton's method
    n = max(2.0, B / 2.0)  # Initial guess near expected range
    for _ in range(1000):
        f  = 2.0*n - B*(math.log(n) + 1)
        fp = 2.0 - B/n
        if abs(fp) < 1e-15: break
        n_new = n - f/fp
        if n_new <= 0: n_new = n/2
        if abs(n_new - n) < 1e-10: break
        n = n_new
    return n

print("Prime helper functions defined.")
print(f"Number of primes ≤ 600: {len(PRIMES_600)}")
print(f"Check: 137 is prime = {137 in PRIMES_600}  ✓")
```

```python
# Step 6: Find n* = 137 from V_eff with B_full

n_cont = continuous_minimum(B_full)
n_star, V_at_nstar = prime_minimizer(B_full)

print("=" * 50)
print("Step 6: Prime Attractor n* = 137  [CONDITIONAL]")
print("=" * 50)
print(f"  B_full = {B_full:.6f}  [MC × MC = double conjecture]")
print(f"  Continuous minimum: n*_cont = √(B_full/2) = {n_cont:.4f}")
print(f"  Prime minimizer:    n*_prime = {n_star}")
print(f"  V_eff(n*) = {V_at_nstar:.4f}")
print()

if n_star == 137:
    print("  ✓ n*_prime = 137 — consistent with α⁻¹_bare = 137")
    print("    STATUS: CONDITIONAL on k=1 (Gap G3-k) AND R ≈ 1.114")
else:
    print(f"  ✗ n*_prime = {n_star} ≠ 137  — check B_full")

print()
print("  V_eff values near n*:")
nearby_primes = [p for p in PRIMES_600 if abs(p - n_star) <= 15]
for p in nearby_primes:
    marker = " ← n*" if p == n_star else ""
    print(f"    V_eff({p:3d}) = {V_eff(float(p), B_full):.4f}{marker}")
```
